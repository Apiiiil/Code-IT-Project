```python
import pandas as pd 
```

```python
df = pd.read_csv("train.csv")
print(df)
```

```python
# print(df.head())
# print(df.tail())
df.info()
print(df.describe())
print("Shape:",df.shape)
print("Columns:",df.columns)
```

```python
## Filtering passengers

older_passengers = df[df["Age"]>30]
print(older_passengers.head())

## Survivors

survivors = df[df["Survived"]==1]
print("Survivors:", survivors.head())

# Expensive tickets

expensive_tickets = df[df["Fare"]>50]
print(expensive_tickets.head())
```

```python
# GroupBy operations

# Avg survival rate by passenger class
survival = df.groupby("Pclass")["Survived"].mean()
print(survival)

# Avg fare by passenger class
fare = df.groupby("Pclass")["Fare"].mean()
print(fare)

# Survival rate grouped by gender
gender_survival = df.groupby("Sex")["Survived"].mean()
print(gender_survival)
```
